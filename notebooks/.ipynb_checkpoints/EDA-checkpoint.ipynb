```python
import pandas as pd
import numpy as np
import os

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score

import xgboost as xgb
import matplotlib.pyplot as plt
import joblib
```

```python
data = pd.read_csv('/content/drive/MyDrive/iafinal/datos.csv')

if 'frame' in data.columns:
    data = data.drop('frame', axis=1)

print("Dimensiones originales:", data.shape)
data.head()
```

```python

def crear_ventanas_con_features(df_raw, window_size=10, stride=5):
    """
    Crea ventanas de tamaño window_size con solapamiento stride para cada label.
    Extrae de cada ventana:
      - Media y varianza de cada landmark (x, y, z).
      - Media y desviación estándar de diferencias (velocidades) entre frames consecutivos.
    Argumentos:
      df_raw: DataFrame con columnas 'label' y 'landmark_{j}_{axis}' para j en 0..32 y axis en {x,y,z}.
      window_size: número de frames por ventana (por defecto 10).
      stride: desplazamiento entre ventanas (por defecto 5).
    Retorna:
      DataFrame con una fila por ventana y sus features.
    """
    n_landmarks = 33
    axes = ['x', 'y', 'z']
    ventanas = []

    for label, group in df_raw.groupby('label'):
        group = group.reset_index(drop=True)
        n_frames = len(group)
        for start in range(0, n_frames - window_size + 1, stride):
            window = group.iloc[start : start + window_size].reset_index(drop=True)
            feats = {'label': label}


            for j in range(n_landmarks):
                for axis in axes:
                    col = f'landmark_{j}_{axis}'
                    vals = window[col].values
                    feats[f'mean_l{j}_{axis}'] = np.mean(vals)
                    feats[f'var_l{j}_{axis}']  = np.var(vals)


            for j in range(n_landmarks):
                for axis in axes:
                    col = f'landmark_{j}_{axis}'
                    vals = window[col].values
                    diffs = np.diff(vals)
                    feats[f'vel_mean_l{j}_{axis}'] = np.mean(diffs)
                    feats[f'vel_std_l{j}_{axis}']  = np.std(diffs)

            ventanas.append(feats)

    df_ventanas = pd.DataFrame(ventanas)
    return df_ventanas
```

```python
expected_cols = ['label'] + [f'landmark_{j}_{axis}' for j in range(33) for axis in ['x','y','z']]
missing = [c for c in expected_cols if c not in data.columns]
if missing:
    raise ValueError(f"Faltan columnas esperadas: {missing}")

window_df = crear_ventanas_con_features(data, window_size=10, stride=5)

print("Shape de window_df:", window_df.shape)
window_df.head()
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(window_df['label'])

X = window_df.drop(columns=['label'])
y = y_encoded


X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("Tamaño X_train:", X_train.shape, "  X_test:", X_test.shape)

import numpy as np
for split_name, labels in zip(['Train', 'Test'], [y_train, y_test]):
    unique, counts = np.unique(labels, return_counts=True)
    print(f"{split_name} distribution:")
    for u, c in zip(unique, counts):
        print(f"  {le.inverse_transform([u])[0]}: {c} muestras")
```

```python
scaler_all = StandardScaler()
X_scaled_all = scaler_all.fit_transform(X)


pca_full = PCA()
pca_full.fit(X_scaled_all)

explained_variance_ratio = pca_full.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance_ratio)

print("Varianza explicada por cada componente (solo > 0.001):")
for i, var_exp in enumerate(explained_variance_ratio):
    if var_exp > 0.001:
        print(f"Componente {i+1}: {var_exp:.4f}")


import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(cumulative_variance) + 1),
         cumulative_variance,
         marker='o', linestyle='--')
plt.title('Varianza explicada acumulada por componentes principales')
plt.xlabel('Número de componentes')
plt.ylabel('Varianza explicada acumulada')
plt.grid(True)
plt.show()
```

```python
pca_2 = PCA(n_components=2)
X_reduced_pca2 = pca_2.fit_transform(X_scaled_all)


unique_labels = np.unique(y)
label_colors = {lbl: plt.get_cmap('tab10')(i) for i, lbl in enumerate(unique_labels)}
colors = [label_colors[val] for val in y]

plt.figure(figsize=(8, 6))
plt.scatter(X_reduced_pca2[:, 0], X_reduced_pca2[:, 1], c=colors, s=15)
plt.title("Distribución (PCA-2D) de ventanas")
for lbl, col in label_colors.items():
    plt.scatter([], [], color=col, label=le.inverse_transform([lbl])[0])
plt.legend()
plt.show()


from sklearn.manifold import TSNE
tsne = TSNE(n_components=2, perplexity=30, learning_rate=200, init='pca', random_state=42)
X_tsne2 = tsne.fit_transform(X_scaled_all)

plt.figure(figsize=(8, 6))
plt.scatter(X_tsne2[:, 0], X_tsne2[:, 1], c=colors, s=15)
plt.title("Distribución (t-SNE) de ventanas")
for lbl, col in label_colors.items():
    plt.scatter([], [], color=col, label=le.inverse_transform([lbl])[0])
plt.legend()
plt.show()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=5, random_state=0, n_init=10)
cluster_labels = kmeans.fit_predict(X_reduced_pca2)

plt.figure(figsize=(8, 6))
plt.scatter(X_reduced_pca2[:, 0], X_reduced_pca2[:, 1], c=cluster_labels, cmap='viridis', s=15)
plt.title("KMeans (5 clusters) sobre PCA-2D")
plt.show()
```

```python
!pip install -q tqdm tqdm-joblib

from sklearn.pipeline import Pipeline
import xgboost as xgb
from sklearn.model_selection import GridSearchCV, ParameterGrid
from tqdm.notebook import tqdm
from tqdm_joblib import tqdm_joblib

pipeline_xgb = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA()),
    ('clf', xgb.XGBClassifier(
        objective='multi:softprob',
        eval_metric='mlogloss',
        tree_method='gpu_hist',
        predictor='gpu_predictor',
        random_state=42,
        verbosity=0
    ))
])


param_grid_xgb = {
    'pca__n_components':     [50, 100],
    'clf__n_estimators':     [100, 200],
    'clf__max_depth':        [6, 9],
    'clf__learning_rate':    [0.01, 0.1],
    'clf__subsample':        [0.8, 1],
    'clf__colsample_bytree': [0.8, 1],
    'clf__reg_alpha':        [0, 0.5],
    'clf__reg_lambda':       [1, 2]
}

grid_xgb = GridSearchCV(
    estimator=pipeline_xgb,
    param_grid=param_grid_xgb,
    cv=3,
    scoring='accuracy',
    verbose=0,
    n_jobs=-1
)

total_combos = len(list(ParameterGrid(param_grid_xgb)))

with tqdm_joblib(tqdm(desc="GridSearch Progress", total=total_combos)):
    grid_xgb.fit(X_train, y_train)

print("Mejores parámetros XGBoost (GPU):")
print(grid_xgb.best_params_)
best_xgb_model = grid_xgb.best_estimator_
```

```python
from sklearn.metrics import accuracy_score, classification_report


scaler = best_xgb_model.named_steps['scaler']
pca    = best_xgb_model.named_steps['pca']
clf    = best_xgb_model.named_steps['clf']


X_test_scaled = scaler.transform(X_test)
X_test_pca    = pca.transform(X_test_scaled)


y_pred_xgb = clf.predict(X_test_pca)


acc_xgb = accuracy_score(y_test, y_pred_xgb)
print(f"Accuracy XGBoost en test: {acc_xgb:.4f}\n")


print("Classification Report XGBoost:")
print(classification_report(
    y_test,
    y_pred_xgb,
    target_names=le.inverse_transform(unique_labels)
))
```

```python
import joblib
joblib.dump(le, 'models/label_encoder.pkl')
joblib.dump(best_xgb_model, 'models/modelo_movimientos.pkl')
scaler = best_xgb_model.named_steps['scaler']
joblib.dump(scaler, 'models/scaler.pkl')

print("LabelEncoder guardado en: models/label_encoder.pkl")
print("Pipeline (scaler + PCA + XGBoost) guardado en: models/modelo_movimientos.pkl")
print("Scaler guardado en: models/scaler.pkl")
```
